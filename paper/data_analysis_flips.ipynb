```python
first_percentages_g = []
second_percentages_g = []

for c in choices:
    first_half = c[:58]
    second_half = c[58:]
    first_percentage = np.sum(first_half)/first_half.shape[0]
    second_percentage = np.sum(second_half)/second_half.shape[0]

    first_percentages_g.append(first_percentage)
    second_percentages_g.append(second_percentage)
```

```python
first_percentages = []
second_percentages = []

for c in choices:
    first_half = c[:58]
    second_half = c[58:]
    first_percentage = np.sum(first_half)/first_half.shape[0]
    second_percentage = np.sum(second_half)/second_half.shape[0]

    first_percentages.append(first_percentage)
    second_percentages.append(second_percentage)
```

```python
param1 = second_percentages
param2 = second_percentages_g

plt.figure(figsize=(15, 8))
x = np.linspace(1,30,30)
plt.scatter(x,param1, label="cafe")
# plt.scatter(x,param2, label="gamble")

for i in range(len(x)):
    # plt.plot([x[i], x[i]], [param1[i], param2[i]], color='gray', linestyle='--', alpha=0.6)
    plt.text(x[i] + 0.3, param1[i], i+1, color='blue', fontsize=8)
    # plt.text(x[i] + 0.3, param2[i], i+1, color='blue', fontsize=8)

plt.legend()
```
